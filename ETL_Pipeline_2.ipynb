```python
import pandas as pd

# Load CSV
orders_df = pd.read_csv("Orders.csv")

# 1. Convert Price column to numeric
orders_df["Price"] = pd.to_numeric(
    orders_df["Price"],
    errors="coerce"
)

# 2. Fill missing Price values with column average
orders_df["Price"] = orders_df["Price"].fillna(
    orders_df["Price"].mean()
)

# 3. Clean Category column
orders_df["Category"] = (
    orders_df["Category"]
    .str.strip()
    .str.lower()
)

# 4. Convert mixed OrderDate formats to datetime
orders_df["OrderDate"] = pd.to_datetime(
    orders_df["OrderDate"],
    format="mixed",
    errors="coerce"
)

# 5. Sort by OrderDate
orders_df = orders_df.sort_values("OrderDate")

# Display result
print(orders_df.head())
```

```python
import pandas as pd

# Load customers CSV
customers_df = pd.read_csv("Customers.csv")

# 1. Clean all string columns
for col in customers_df.select_dtypes(include="object").columns:
    customers_df[col] = customers_df[col].str.strip()

# 2. Convert Category to lowercase if it exists
if "Category" in customers_df.columns:
    customers_df["Category"] = customers_df["Category"].str.lower()

# 3. Convert date column if it exists
if "OrderDate" in customers_df.columns:
    customers_df["OrderDate"] = pd.to_datetime(
        customers_df["OrderDate"],
        format="mixed",
        errors="coerce"
    )

# 4. Sort by date if OrderDate exists
if "OrderDate" in customers_df.columns:
    customers_df = customers_df.sort_values("OrderDate")

# 5. Display cleaned customer data
print(customers_df.head())
```

```python
import pandas as pd

# Load customers data
customers_df = pd.read_csv("Customers.csv")

# Merge orders and customers using CustomerID
merged_df = pd.merge(
    orders_df,
    customers_df,
    on="CustomerID",
    how="inner"
)

print(merged_df.head())
```

```python
# Save the merged DataFrame as a CSV file
merged_df.to_csv("merged_orders_customers.csv", index=False)

print("Merged data saved to merged_orders_customers.csv")
```

```python
# Calculate revenue for each order and display total revenue by country
# in descending order of revenue.

# Create Revenue column
merged_df["Revenue"] = merged_df["Price"] * merged_df["Quantity"]

# Calculate total revenue by country and sort in descending order
revenue_by_country = (
    merged_df.groupby("Country", as_index=False)["Revenue"]
    .sum()
    .sort_values("Revenue", ascending=False)
)

print(revenue_by_country)
```
